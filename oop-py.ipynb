```python
#in this section we are goin to understand python opp dunder methods
#dunder methods are special methods in python that have double underscores before 
# and after their names. they are also known as magic methods or special methods. 
# these methods are used to define the behavior of objects in python. 
# they allow us to implement operator overloading, 
# which means we can define how operators like +, -, *, /, etc. work with our custom objects.
```

```python
#__repr__ -- cómo se ve tu objeto
class Perro:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def __repr__(self):
        return f"Perro(nombre='{self.nombre}', edad={self.edad})"

p = Perro('Rex', 3)
print(p)
repr(p)

#sin __repr__ el resultado sería algo como <__main__.Perro object at 0x7f8b9c2d0>, que no es muy informativo. Con __repr__, obtenemos una representación clara y útil del objeto.
```

```python
#__add__ -- cómo se comporta tu objeto con el operador +
#cuando escribes a + b, Python internamente llama a a.__add__(b). Esto te permite definir cómo se suman tus objetos
#tu decides que significa "sumar" para tu clase
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

v1 = Vector(1, 2)
v2 = Vector(3, 4)
v3 = v1 + v2 #python llama a v1.__add__(v2)
v3

#sin __add__, el operador + no funcionaría para tus objetos, y obtendrías un error si intentas usarlo. Con __add__, puedes definir exactamente cómo quieres que se comporten tus objetos cuando se suman.
#en este caso, como vectores, tiene sentido que la suma de dos vectores sea un nuevo vector con las componentes sumadas.
```

```python
#__mul__ -- cómo se comporta tu objeto con el operador *
#similar a __add__, pero para el operador *

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __mul__(self, scalar):
        return Vector(self.x * scalar, self.y * scalar)

v = Vector(2, 3)
v * 4 #python llama a v.__mul__(4)

#v * 4 funciona, pero 4 * v no -- para eso existe __rmul__, 
#Python llama cuando el objeto está ala derecha del operador
```

```python
#La idea central de OOP
#En lugar de escribir funciones sueltas que operan sobre datos
#en OOP agrupas los datos y las funciones juntos dentro de una clase

# Sin OOP -- datos y funciones separados
nombre = 'Rex'
edad = 3
def ladrar(nombre):
    print (f"{nombre} dice: ¡Guau!")

# Con OOP -- datos y funciones juntos en un objeto
class Perro:
    def __init__(self, nombre, edad): # constructor
        self.nombre = nombre          # atributo
        self.edad = edad

    def ladrar(self):                  # método
        print(f"{self.nombre} dice: ¡Guau!")

rex = Perro('Rex', 3)   #crear un objeto (instancia)
rex.ladrar()            #llamar al método ladrar del objeto rex
```

```python
# ¿Donde encajan las dunder methods?
# Son parte de la encapsulación -- te permiten controlar cómo Python
# trata a tus objetos con operadores y funciones built-in

class Perro:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def __repr__(self):             # Controla print()
        return f"Perro(nombre='{self.nombre}', edad={self.edad})"

    def __add__(self, other):
        return Perro(self.nombre + ' & ' + other.nombre, self.edad + other.edad)
    
    def __eq__(self, other):        # Controla ==, compara los nombres de los perros
        return self.nombre == other.nombre

rex1 = Perro('Rex', 3)
rex2 = Perro('Rex', 5)
print(rex1)          # Usa __repr__
print(rex1 + rex2)   # Usa __add__  
print(rex1 == rex2)  # Usa __eq__, compara nombres

# Son simplemente métodos especiales que Python llama en situaciones específicas.
# No son mágicos, son parte de la forma en que Python maneja objetos y operadores
    
```

```python
# Self representa el objeto mismo, es como "yo" en el contexto de la clase.
# Es el primer parámetro de los métodos de instancia, y se usa para acceder a los
# atributos y otros métodos del objeto. Es una convención usar "self", pero podrías usar cualquier nombre.

class Perro:
    def __init__(self, nombre):
        self.nombre = nombre        # "mi propio nombre es---"
    
    def ladrar(self):
        print(f"{self.nombre} dice: ¡Guau!") # "mi nombre es--- dice: ¡Guau!"

rex = Perro('Rex')
rex.ladrar() 

# Cuando escribes rex.ladrar(), Python internamente hace algo como
Perro.ladrar(rex) # llama al método ladrar de la clase Perro, pasando el objeto rex como el primer argumento (self)
```
